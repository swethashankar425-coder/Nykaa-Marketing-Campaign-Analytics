##KPI CREATIONS

CTR

```
data["CTR"] = (data["Clicks"] / data["Impressions"]) * 100
print (data["CTR"])
```

CONVERATION RATE

```
data["Conversion_Rate"] = (data["Conversions"] / data["Clicks"]) * 100
print (data["Conversion_Rate"])
```

AD SPEND

```
data["Ad_Spend"] = data["Acquisition_Cost"] * data["Conversions"]
print (data["Ad_Spend"])
```

CPC

```
data["CPC"] = data["Ad_Spend"] / data["Clicks"]
print (data["CPC"])
```

CPL

```
data["CPL"] = data["Ad_Spend"] / data["Leads"]
print (data["CPL"])
```
